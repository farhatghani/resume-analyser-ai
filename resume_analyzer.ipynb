```python
# 
from pdfminer.high_level import extract_text
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import re
import docx
```

```python
# 
def extract_text(file_path):
    text = extract_text(file_path)
    return text
```

```python
# 
def extract_docx(file_path):
    doc = docx.Document(file_path)
    text = "\n".join([para.text for para in doc.paragraphs])
    return text
```

```python
# 
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]',' ', text)
    text = re.sub(r'\s+',' ', text)
    return text
    
```

```python
# 
resume_text = extract_text(r"C:\Users\farha\Documents\Resume\sample_resume.pdf")
print (resume_text[:1000])
```

```python
# 
skills = [
    "python",
    "machine learning",
    "deep learning",
    "sql",
    "docker",
    "aws"
]
def extract_skills(text):
    found = []
    text = text.lower()
    for skill in skills:
        if skill in text:
            found.append(skill)

    return found
print (extract_skills(resume_text))
```

```python
# 
resume_text = extract_text(r"C:\Users\farha\Documents\Resume\Farhat_Ghani_resume_DA.pdf")
resume_text = clean_text(resume_text)
job_text = """
Looking for a Data analyst with python, SQL, PowerBi and machine learning experience."""
job_text = clean_text(job_text)
```

```python
print(job_text.lower())
```

```python
print(repr(job_text[:500]))
```

```python
resume_skills = extract_skills(resume_text)
print(resume_skills)
```

```python
job_skills = extract_skills(job_text)
print(job_skills)
```

```python
missing_skills = list(set(job_skills) -
                      set(resume_skills))
print(missing_skills)
```

```python
matched = len(set(job_skills) & 
            set(resume_skills))
total = len(job_skills)
score = (matched / total) * 100
print(f"Match Score:{score:.2f}%")
```

```python


documents = [resume_text, job_text]
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform (documents)
similarity = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

score = similarity[0][0] * 100
print (f"overall Match: {score:2f}%")
```

```python
resume_skills = extract_skills(resume_text)
job_skills = extract_skills(job_text)

print("resume_skills:", resume_skills)
print("job_skills:", job_skills)

matched = set(resume_skills) & set(job_skills)
missing = set(job_skills) - set(resume_skills)

print("Matched skills:", matched)
print("Missing skills:", missing)
```

```python
skill_score = ( len(matched) / len(job_skills)) * 100
```

```python
# Printing overall result of the risume scanning
print(f"Overall Similarity: {score:.2f}%")
print()

print(f"Skills Match: {skill_score:.0f}%")
print()

print("Matched Skills:")
for skill in matched:
    print(f"✓ {skill}")

print()

print("Missing Skills:")
for skill in missing:
    print(f"✗ {skill}")

print()

if skill_score >= 80:
    recommendation = "Highly Suitable"
elif skill_score >= 60:
    recommendation = "Suitable"
elif skill_score >= 40:
    recommendation = "Moderately Suitable"
else:
    recommendation = "Not Suitable"

print(f"Recommendation: {recommendation}")
```
